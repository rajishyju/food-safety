```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
df=pd.read_csv('food_safety.csv')
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=pd.get_dummies(df,columns=["product_type"],drop_first=True)
df=pd.get_dummies(df,columns=["packaging_type"],drop_first=True)
df
```

```python
df['product_approval'].value_counts()
df["product_approval"]=df["product_approval"].map({"Approved":1,"Rejected":0})
df.head()
```

```python
x=df.drop("product_approval",axis=1)
y=df["product_approval"]
x
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=MinMaxScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_sclaled=scaler.transform(x_test)
```

```python
x_train_scaled=pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_test_sclaled=pd.DataFrame(x_test_sclaled,columns=x_test.columns)
```

```python
models = {
"Logistic Regression": LogisticRegression(max_iter=1000),
"Naive Bayes": GaussianNB(),
"Decision Tree": DecisionTreeClassifier(random_state=42),
"Random Forest": RandomForestClassifier(random_state=42)
}
models
```

```python
results={}
for name,model in models.items():
    print(f"\n====={name}=====")
    if name in ["Logistic Regression","Naive Bayes"]:
                model.fit(x_train_scaled,y_train)
                y_pred=model.predict(x_test_sclaled)  
    else:
        model.fit(x_train,y_train)
        y_pred=model.predict(x_test)

    acc=accuracy_score(y_test,y_pred)
    pre=precision_score(y_test,y_pred)
    rec=recall_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)
    results[name]=[acc,pre,rec,f1]
    print("Accuracy:",acc)
    print("precision",pre)
    print("recall",rec)
    print("f1score",f1)
    #results_df=pd.DataFrame(results,index=["Accuracy","precision","recall","f1_score"]).T
    results_df = pd.DataFrame(
    results, 
    index=["Accuracy", "Precision", "Recall", "F1_Score"]
).T
    results_df

    
               
```

```python
# Checking best ML model for the data
best_model_name = results_df["recall"].idxmax()
print("Best model based on Recall:", best_model_name)
best_model = models[best_model_name]

if best_model_name in ["Logistic Regression", "Naive Bayes"]:
    best_model.fit(x_train_scaled, y_train)
else:
    best_model.fit(x_train, y_train)

# Saving the model
with open("model.pkl", "wb") as f:
    pickle.dump(best_model, f)

# Saving the scaler
if best_model_name in ["Logistic Regression", "Naive Bayes"]:
    with open("scaler.pkl", "wb") as f:
        pickle.dump(scaler, f)
```
